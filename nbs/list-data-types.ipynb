```python
import requests
from pprint import pprint
path= "/api/events/"
base_url = "http://localhost:8000"
endpoint = f"{base_url}{path}"
endpoint
```

```python
response = requests.get(endpoint)
print(response.status_code)
if response.ok:
    data = response.json()
    pprint(data)
```
